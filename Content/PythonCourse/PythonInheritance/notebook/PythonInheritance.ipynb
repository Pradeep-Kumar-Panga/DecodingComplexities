## The Copy-Paste Problem

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"{self.name} is an animal"

class Cat:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"{self.name} is an animal"

print(Dog("Rex").describe())
print(Cat("Tom").describe())
```

## A Parent Class and a Child Class

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"{self.name} is an animal"

    def speak(self):
        return "..."

class Dog(Animal):
    pass

rex = Dog("Rex")

print(rex.describe())
print(rex.speak())
```

## Override What Is Different

```python
class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

print(Dog("Rex").speak())
print(Cat("Tom").speak())
print(Animal("Generic").speak())
```

## Extend the Parent With super()

```python
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def speak(self):
        return "Woof"

rex = Dog("Rex", "Labrador")

print(rex.name, rex.breed)
print(rex.describe())
```

## Checking Types

```python
print(isinstance(rex, Dog))
print(isinstance(rex, Animal))
print(issubclass(Dog, Animal))
print(isinstance(rex, Cat))
```

## One Call, Many Behaviors

```python
pets = [Dog("Rex", "Labrador"), Cat("Tom")]

for pet in pets:
    print(pet.name, "says", pet.speak())
```
