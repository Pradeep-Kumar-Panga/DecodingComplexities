## Python Data to JSON Text

```python
import json

person = {
    "name": "Asha",
    "age": 31,
    "skills": ["python", "sql"],
    "active": True,
    "manager": None,
}

text = json.dumps(person)

print(text)
print(type(text))
```

## JSON Text Back to Python Data

```python
data = json.loads(text)

print(data["skills"][1])
print(type(data))
print(data == person)
```

## Make It Readable

```python
print(json.dumps(person, indent=2, sort_keys=True))
```

## Save to a File and Load It Back

```python
with open("person.json", "w") as file:
    json.dump(person, file, indent=2)

with open("person.json") as file:
    loaded = json.load(file)

print(loaded == person)
with open("person.json") as file:
    print(file.read())
```

## What JSON Cannot Keep Exactly

```python
print(json.dumps((1, 2)))
print(json.loads(json.dumps((1, 2))))
print(json.dumps({1: "a"}))
```

```python
json.dumps({"tags": {"a", "b"}})
```

## When the Text Is Not Valid JSON

```python
json.loads("{'name': 'Asha'}")
```
