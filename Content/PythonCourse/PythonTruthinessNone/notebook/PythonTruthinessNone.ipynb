## Every Value Can Be Tested

```python
values = [0, 1, -1, 0.0, "", "a", " ", [], [0], {}, (), None]

for value in values:
    print(repr(value), bool(value))
```

## Using Truthiness in if Statements

```python
items = []

if items:
    print("There are items to process")
else:
    print("Nothing to process")

items.append("pen")

if items:
    print("There are items to process")
else:
    print("Nothing to process")
```

## None Means "No Value"

```python
result = None
print(result)
print(type(result))
print(bool(result))
```

```python
print(None == 0)
print(None == "")
print(0 == "")
```

## Test for None With is

```python
score = None
print(score is None)

score = 0
print(score is None)
```

```python
score = 0

if not score:
    print("No score yet")

if score is None:
    print("No score yet (checked with is None)")
else:
    print("The score is", score)
```
