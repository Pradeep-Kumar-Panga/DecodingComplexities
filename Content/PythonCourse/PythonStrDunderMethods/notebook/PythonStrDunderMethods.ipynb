## The Unhelpful Default

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(2, 3)
print(p)
```

## __str__ Decides What print() Shows

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

p = Point(2, 3)

print(p)
print(f"The point is {p}")
```

## __repr__ Is for Developers

```python
print([Point(1, 2), Point(3, 4)])
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

print(Point(2, 3))
print([Point(1, 2), Point(3, 4)])
```

## __len__ Makes len() Work

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

mix = Playlist(["song a", "song b", "song c"])
print(len(mix))
```

## __eq__ Defines What "Equal" Means

```python
print(Point(1, 2) == Point(1, 2))

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

print(Point(1, 2) == Point(1, 2))
print(Point(1, 2) == Point(5, 6))
```

## __add__ Teaches + a New Trick

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

print(Point(1, 2) + Point(3, 4))
```
