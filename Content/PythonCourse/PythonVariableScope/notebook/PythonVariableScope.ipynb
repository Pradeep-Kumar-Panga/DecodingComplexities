## Variables Born Inside a Function Stay There

```python
def make_total():
    total = 10 + 5
    return total

print(make_total())
print(total)
```

## Functions Can Read Outside Variables

```python
rate = 0.25

def add_tax(price):
    return price + price * rate

print(add_tax(200))
```

## Assigning Inside a Function Creates a New Local

```python
counter = 0

def reset():
    counter = 99
    print("inside:", counter)

reset()
print("outside:", counter)
```

```python
def increase():
    counter = counter + 1

increase()
```

## The global Keyword, and a Better Way

```python
counter = 0

def increase():
    global counter
    counter = counter + 1

increase()
increase()
print(counter)
```

```python
def increase_by_one(value):
    return value + 1

counter = 0
counter = increase_by_one(counter)
counter = increase_by_one(counter)
print(counter)
```

## Functions Inside Functions

```python
def outer():
    message = "from outer"

    def inner():
        print(message)

    inner()

outer()
```

## One More Gotcha: Hiding a Built-in Name

```python
list = [1, 2, 3]
list("abc")
```

```python
del list
print(list("abc"))
```
