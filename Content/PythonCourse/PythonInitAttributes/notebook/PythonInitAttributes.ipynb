## The Fragile Way

```python
class Dog:
    pass

rex = Dog()
print(rex.name)
```

## __init__ Runs When You Create an Object

```python
class Dog:
    def __init__(self, name, age):
        print("creating", name)
        self.name = name
        self.age = age

rex = Dog("Rex", 3)
fido = Dog("Fido", 5)

print(rex.name, rex.age)
print(fido.name, fido.age)
```

## What self.name = name Means

```python
class Cat:
    def __init__(self, name):
        name = name

tom = Cat("Tom")
print(tom.name)
```

## Default Values

```python
class Dog:
    def __init__(self, name, age=1):
        self.name = name
        self.age = age

puppy = Dog("Bolt")
print(puppy.name, puppy.age)
```

## Instance Attributes and Class Attributes

```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name):
        self.name = name

rex = Dog("Rex")
fido = Dog("Fido")

print(Dog.species, rex.species, fido.species)

rex.species = "wolf"
print(rex.species, fido.species, Dog.species)

Dog.species = "dog"
print(fido.species, rex.species)
```

## The Shared List Trap

```python
class Dog:
    tricks = []

    def __init__(self, name):
        self.name = name

rex = Dog("Rex")
fido = Dog("Fido")

rex.tricks.append("sit")
print(fido.tricks)
```

```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.tricks = []

rex = Dog("Rex")
fido = Dog("Fido")

rex.tricks.append("sit")
print(rex.tricks)
print(fido.tricks)
```
