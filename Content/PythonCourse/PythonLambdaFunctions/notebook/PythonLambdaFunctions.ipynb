## A Function You Only Need Once

```python
words = ["banana", "fig", "apple", "kiwi"]

def by_length(word):
    return len(word)

print(sorted(words, key=by_length))
```

## The lambda Syntax

```python
print(sorted(words, key=lambda word: len(word)))
```

```python
square = lambda n: n * n
add = lambda a, b: a + b

print(square(7))
print(add(3, 4))
```

## Sorting Records by One Field

```python
people = [("Asha", 31), ("Ravi", 25), ("Meena", 38)]

print(sorted(people, key=lambda person: person[1]))
print(sorted(people, key=lambda person: person[1], reverse=True))
```

## With map() and filter()

```python
numbers = [1, 2, 3, 4, 5, 6]

print(list(map(lambda n: n * 10, numbers)))
print(list(filter(lambda n: n % 2 == 0, numbers)))
```

## The Limits of lambda

```python
label = lambda n: "even" if n % 2 == 0 else "odd"

print(label(6))
print(label(7))
```
