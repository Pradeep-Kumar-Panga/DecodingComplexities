## Create a Virtual Environment

```python
import os
import subprocess
import sys

subprocess.run([sys.executable, "-m", "venv", "demo_env"], check=True)
print(sorted(os.listdir("demo_env")))
```

## Prove It Is Separate

```python
result = subprocess.run(
    ["demo_env/bin/python", "-c", "import sys; print(sys.prefix != sys.base_prefix)"],
    capture_output=True,
    text=True,
)
print(result.stdout.strip())
```
