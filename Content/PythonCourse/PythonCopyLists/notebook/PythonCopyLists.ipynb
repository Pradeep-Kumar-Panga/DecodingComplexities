## The Surprise

```python
a = [1, 2, 3]
b = a
b.append(4)

print(a)
print(b)
```

```python
print(a is b)
```

## Make a Real Copy

```python
c = a.copy()
d = list(a)
e = a[:]

c.append(5)

print(a)
print(c)
print(c is a, d is a, e is a)
```

## When a Copy Still Shares Things

```python
matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()

shallow[0].append(99)

print(matrix)
print(shallow)
print(shallow[0] is matrix[0])
```

## Deep Copy for Nested Data

```python
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[1].append(55)

print(original)
print(deep)
```

## The Same Trap With Function Arguments

```python
def add_item(items):
    items.append("extra")

shopping = ["bread", "milk"]
add_item(shopping)

print(shopping)
```

```python
def with_item(items):
    new_items = items.copy()
    new_items.append("extra")
    return new_items

shopping = ["bread", "milk"]
bigger = with_item(shopping)

print(shopping)
print(bigger)
```
