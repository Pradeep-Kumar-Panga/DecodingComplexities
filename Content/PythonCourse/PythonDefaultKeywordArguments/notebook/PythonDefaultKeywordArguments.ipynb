## The Problem: Repeating the Same Value

```python
def describe(name, city, country):
    return name + " lives in " + city + ", " + country

print(describe("Asha", "Pune", "India"))
print(describe("Ravi", "Delhi", "India"))
```

## Default Values

```python
def describe(name, city, country="India"):
    return name + " lives in " + city + ", " + country

print(describe("Asha", "Pune"))
print(describe("Meena", "Paris", "France"))
```

```python
def broken(a=1, b):
    return a + b
```

## Keyword Arguments

```python
print(describe(city="Pune", name="Asha"))
print(describe("Ravi", country="Nepal", city="Kathmandu"))
```

## The Mutable Default Trap

```python
def add_task(task, tasks=[]):
    tasks.append(task)
    return tasks

print(add_task("write"))
print(add_task("test"))
```

```python
def add_task(task, tasks=None):
    if tasks is None:
        tasks = []
    tasks.append(task)
    return tasks

print(add_task("write"))
print(add_task("test"))
```
