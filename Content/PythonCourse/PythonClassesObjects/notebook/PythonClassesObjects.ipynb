## The Problem With Loose Data

```python
dog1 = {"name": "Rex", "age": 3}
dog2 = {"name": "Fido", "age": 5}

def describe(dog):
    return dog["name"] + " is " + str(dog["age"])

print(describe(dog1))
print(describe(dog2))
```

## Define a Class and Create Objects

```python
class Dog:
    pass

rex = Dog()

print(type(rex))
print(isinstance(rex, Dog))
```

## Objects Hold Their Own Data

```python
rex.name = "Rex"
rex.age = 3

fido = Dog()
fido.name = "Fido"

print(rex.name, rex.age)
print(fido.name)
```

```python
print(fido.age)
```
