## The Index-Juggling Way

```python
names = ["Asha", "Ravi", "Meena"]
scores = [82, 67, 91]

for i in range(len(names)):
    print(names[i], scores[i])
```

## zip() Pairs Items Up

```python
for name, score in zip(names, scores):
    print(name, score)
```

```python
print(list(zip(names, scores)))
```

```python
print(zip(names, scores))
```

## Three or More Lists

```python
passed = [True, True, False]

for name, score, ok in zip(names, scores, passed):
    print(name, score, ok)
```

## When the Lists Are Different Lengths

```python
print(list(zip([1, 2, 3], ["a", "b"])))
```

```python
print(list(zip([1, 2, 3], ["a", "b"], strict=True)))
```

## Build a Dictionary From Two Lists

```python
score_by_name = dict(zip(names, scores))
print(score_by_name)
```

## Unzip: Split Pairs Back Into Lists

```python
pairs = [("a", 1), ("b", 2), ("c", 3)]
letters, numbers = zip(*pairs)

print(letters)
print(numbers)
```
