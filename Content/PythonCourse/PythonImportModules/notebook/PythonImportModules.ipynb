## Some Tools Are Not Loaded by Default

```python
print(sqrt(16))
```

```python
import math

print(math.sqrt(16))
print(math.pi)
print(math.factorial(5))
```

## Three Ways to Import

```python
import math
from math import sqrt, pi
import statistics as stats

print(math.sqrt(25))
print(sqrt(25))
print(pi)
print(stats.mean([2, 4, 6, 8]))
```

## Make Your Own Module

```python
%%writefile greetings.py
def hello(name):
    return f"Hello, {name}!"

GREETING_COUNT = 2
```
