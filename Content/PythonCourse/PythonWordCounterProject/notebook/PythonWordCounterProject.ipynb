## The Input: A Small Text File

```python
%%writefile story.txt
The quick brown fox jumps over the lazy dog. The dog barks, and the fox runs away.
Quick thinking saves the fox, but the dog never forgets. A fox is quick; a dog is loyal.
```

## Step 1: Read the File and Split It Into Words

```python
with open("story.txt") as file:
    text = file.read()

words = text.split()

print(len(words))
print(words[:6])
```

## Step 2: Clean the Words

```python
cleaned = [word.strip(".,;:!?").lower() for word in words]

print(cleaned[:8])
```

## Step 3: Count With a Dictionary

```python
counts = {}

for word in cleaned:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```

## Step 4: Sort and Show the Top Five

```python
ranked = sorted(counts.items(), key=lambda item: item[1], reverse=True)

for word, count in ranked[:5]:
    print(f"{word:<8}{count:>3}")
```

## Step 5: Tidy It Into Functions

```python
def read_words(path):
    with open(path) as file:
        return [word.strip(".,;:!?").lower() for word in file.read().split()]

def count_words(words):
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    return counts

def top_words(counts, n=5):
    ranked = sorted(counts.items(), key=lambda item: item[1], reverse=True)
    return ranked[:n]

for word, count in top_words(count_words(read_words("story.txt"))):
    print(f"{word:<8}{count:>3}")
```

## Step 6: Handle a Missing File

```python
try:
    read_words("missing.txt")
except FileNotFoundError:
    print("File not found: missing.txt")
```

## Step 7: Save the Result as JSON

```python
import json

result = dict(top_words(count_words(read_words("story.txt"))))

with open("top_words.json", "w") as file:
    json.dump(result, file, indent=2)

with open("top_words.json") as file:
    print(file.read())
```

## Bonus: Python Already Has a Counter

```python
from collections import Counter

print(Counter(read_words("story.txt")).most_common(5))
print(Counter(read_words("story.txt")).most_common(5) == top_words(count_words(read_words("story.txt"))))
```
