## Ask a Question and Keep the Answer

```python
name = input("What is your name? ")
print("Hello,", name)
```

## The Answer Is Always Text

```python
age = input("How old are you? ")
print(age)
print(type(age))
```

```python
print(age + 1)
```

## Convert the Answer With int() or float()

```python
age = int(age)
print(age + 1)
```

```python
price = float(input("Price: "))
print(price * 2)
```

## When the Answer Is Not a Number

```python
int("forty")
```

```python
answer = input("How many tickets? ")
print(answer.isdigit())
print("40".isdigit())
```

## Putting It Together

```python
name = input("Name: ")
first = int(input("First number: "))
second = int(input("Second number: "))

print(name, "your total is", first + second)
```
