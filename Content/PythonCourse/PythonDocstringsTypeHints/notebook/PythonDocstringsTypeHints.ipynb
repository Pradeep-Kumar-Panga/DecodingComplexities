## The Mystery Function

```python
def calc(a, b, c):
    return a * b * (1 + c / 100)

print(calc(200, 3, 25))
```

## Add a Docstring

```python
def total_price(unit_price, quantity, tax_percent):
    """Return the final price including tax.

    unit_price: price of one item
    quantity: number of items
    tax_percent: tax as a percentage, for example 25 for 25%
    """
    return unit_price * quantity * (1 + tax_percent / 100)

print(total_price(200, 3, 25))
```

```python
print(total_price.__doc__)
```

## Add Type Hints

```python
def total_price(unit_price: float, quantity: int, tax_percent: float) -> float:
    """Return the final price including tax."""
    return unit_price * quantity * (1 + tax_percent / 100)

print(total_price(200, 3, 25))
```

```python
print(total_price.__annotations__)
```

## Hints Do Not Stop Wrong Calls

```python
print(total_price(200, "3", 25))
```

## Hints for Lists and Optional Values

```python
def first_word(sentence: str) -> str | None:
    """Return the first word, or None when the sentence is empty."""
    words = sentence.split()
    if not words:
        return None
    return words[0]

def total(numbers: list[int]) -> int:
    """Return the sum of a list of whole numbers."""
    return sum(numbers)

print(first_word("hello brave world"))
print(first_word(""))
print(total([4, 5, 6]))
```
