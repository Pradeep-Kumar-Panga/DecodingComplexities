## The Limit of a Fixed List of Parameters

```python
def add(a, b):
    return a + b

add(1, 2, 3)
```

## *args Collects Extra Positional Arguments

```python
def add_all(*numbers):
    print(numbers)
    return sum(numbers)

print(add_all(1, 2, 3))
print(add_all(10, 20))
print(add_all())
```

## **kwargs Collects Extra Keyword Arguments

```python
def show_profile(**details):
    for key, value in details.items():
        print(key, "=", value)

show_profile(name="Asha", city="Pune", age=31)
```

## Mixing Them With Normal Parameters

```python
def log(level, *messages, **options):
    separator = options.get("sep", " ")
    print(level + ":", separator.join(messages))

log("INFO", "server", "started")
log("WARN", "disk", "almost", "full", sep="-")
```

## The Stars When Calling a Function

```python
values = [3, 4, 5]
print(add_all(*values))

profile = {"name": "Ravi", "age": 29}
show_profile(**profile)
```

## A Practical Use: Passing Everything Along

```python
def loud_print(*args, **kwargs):
    print("!!!", *args, **kwargs)

loud_print("hello", "world", sep="_")
```
