## Two Values That Look Alike but Are Not

```python
text_five = "5"
number_five = 5

print(text_five + text_five)
print(number_five + number_five)
print(type(text_five), type(number_five))
```

```python
print(text_five + number_five)
```

## int(), float() and str()

```python
print(int("5") + 5)
print(text_five + str(5))
print(float("3.5") + 1)
```

```python
print(int(3.9))
print(int(-3.9))
print(round(3.9))
```
