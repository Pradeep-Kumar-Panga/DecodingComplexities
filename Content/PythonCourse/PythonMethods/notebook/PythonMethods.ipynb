## A Class With Methods

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def summary(self):
        return f"{self.owner}: {self.balance}"

acct = Account("Asha", 100)
acct.deposit(50)

print(acct.balance)
print(acct.summary())
```

## What Python Does Behind the Scenes

```python
Account.deposit(acct, 25)
print(acct.balance)

acct.deposit(25)
print(acct.balance)
```

## Methods That Change Data and Methods That Return It

```python
result = acct.deposit(10)

print(result)
print(acct.balance)
```

## Methods Can Call Other Methods

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def apply_interest(self, rate):
        self.deposit(self.balance * rate)

savings = Account("Ravi", 200)
savings.apply_interest(0.25)

print(savings.balance)
```

## Protecting the Object From Bad Requests

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount

wallet = Account("Meena", 80)
wallet.withdraw(500)
```

```python
print(wallet.balance)
wallet.withdraw(30)
print(wallet.balance)
```

## Each Object Keeps Its Own Data

```python
first = Account("Asha", 100)
second = Account("Ravi", 100)

first.deposit(40)

print(first.balance, second.balance)
```
