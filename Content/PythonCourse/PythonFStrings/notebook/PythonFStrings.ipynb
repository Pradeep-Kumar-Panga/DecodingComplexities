## The Glued-Together Way

```python
name = "Asha"
age = 31

print("Name: " + name + ", age: " + str(age))
```

```python
print("Name: " + name + ", age: " + age)
```

## Write the Message With an f-string

```python
print(f"Name: {name}, age: {age}")
```

```python
print(f"Next year {name.upper()} will be {age + 1}")
```

```python
print(f"{age=}")
```

## Format Numbers

```python
price = 1234.5678

print(f"{price:.2f}")
print(f"{1234567:,}")
print(f"{0.256:.1%}")
print(f"{42:05d}")
```

## Line Up Columns

```python
rows = [("pen", 3), ("notebook", 12), ("stapler", 1)]

for item, quantity in rows:
    print(f"{item:<10}{quantity:>4}")
```

## The Older format() Method

```python
print("{} is {} years old".format(name, age))
print("{:.2f}".format(price))
```

## Quotes and Curly Braces Inside f-strings

```python
data = {"city": "Pune"}
print(f"City: {data['city']}")
```

```python
print(f"Use {{braces}} for placeholders, like {name}")
```
