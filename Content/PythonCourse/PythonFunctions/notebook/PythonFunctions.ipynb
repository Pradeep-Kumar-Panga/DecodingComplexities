## The Problem: The Same Code, Three Times

```python
price_a = 200
total_a = price_a + price_a * 0.18
print(total_a)

price_b = 350
total_b = price_b + price_b * 0.18
print(total_b)

price_c = 1000
total_c = price_c + price_c * 0.18
print(total_c)
```

## Write It Once with def

```python
def add_tax(price):
    total = price + price * 0.18
    return total
```

```python
print(add_tax(200))
print(add_tax(350))
print(add_tax(1000))
```

## Parameters and Arguments

```python
def add_tax(price, rate):
    return price + price * rate

print(add_tax(200, 0.18))
print(add_tax(200, 0.05))
```

```python
add_tax(200)
```

## return Sends a Value Back

```python
def add_tax_print_only(price):
    print(price + price * 0.18)

result = add_tax_print_only(200)
print("result is:", result)
```

```python
def low_and_high(numbers):
    return min(numbers), max(numbers)

low, high = low_and_high([14, 3, 27, 9])
print(low)
print(high)
```
