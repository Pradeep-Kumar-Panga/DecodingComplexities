## Create a Date

```python
from datetime import date, datetime, timedelta

d = date(2026, 3, 5)

print(d)
print(d.year, d.month, d.day)
print(d.weekday())
print(d.strftime("%A"))
```

## Subtract Two Dates

```python
gap = date(2026, 3, 5) - date(2026, 2, 10)

print(gap)
print(gap.days)
```

## Add and Subtract Time With timedelta

```python
print(d + timedelta(days=30))
print(d - timedelta(weeks=2))
```

```python
meeting = datetime(2026, 3, 5, 14, 30)
later = meeting + timedelta(hours=10)

print(meeting)
print(later)
print((later - meeting).total_seconds())
```

## Format a Date as Text

```python
print(meeting.strftime("%d %B %Y, %H:%M"))
print(meeting.strftime("%d/%m/%y"))
print(meeting.strftime("%A"))
```

## Read a Date From Text

```python
christmas = datetime.strptime("25/12/2026", "%d/%m/%Y")

print(christmas)
print(christmas.strftime("%A"))
```

```python
datetime.strptime("2026-12-25", "%d/%m/%Y")
```

## Compare Dates

```python
deadline = date(2026, 3, 31)

print(d < deadline)
print(sorted([date(2026, 5, 1), date(2026, 1, 15), date(2026, 3, 5)]))
```
