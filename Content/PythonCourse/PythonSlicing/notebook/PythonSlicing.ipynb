## Indexes Come First

```python
word = "Python"

print(word[0])
print(word[5])
print(word[-1])
print(word[-2])
```

## The [start:stop] Slice

```python
print(word[0:2])
print(word[2:5])
print(len(word[2:5]))
```

## Leaving Parts Out

```python
print(word[:3])
print(word[3:])
print(word[:])
print(word[-3:])
```

## The Step

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[::2])
print(numbers[1::2])
print(numbers[::-1])
print(numbers)
```

## Slices Work on Tuples Too

```python
coordinates = (1, 2, 3, 4, 5)
print(coordinates[1:4])
```

## Slices Are Copies and Forgive Out-of-Range Limits

```python
part = numbers[1:3]
part[0] = 99
print(part)
print(numbers)
```

```python
print(word[2:100])
print(word[100])
```

## A Practical Example: File Names and Phone Numbers

```python
filename = "report_final.pdf"
print(filename[-3:])
print(filename[:-4])

phone = "9876543210"
print("XXXXXX" + phone[-4:])
```
